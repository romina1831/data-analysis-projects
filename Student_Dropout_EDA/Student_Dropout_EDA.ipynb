# معرفی دیتاست پیش‌بینی ترک تحصیل و موفقیت تحصیلی دانشجویان

## توضیح کلی

این دیتاست در چارچوب یک پروژه‌ی پژوهشی در پرتغال ساخته شده که هدفش کمک به **کاهش افت تحصیلی و ترک تحصیل در آموزش عالی** است. ایده‌ی اصلی این است که با استفاده از تکنیک‌های یادگیری ماشین، دانشجویان در معرض خطر ترک تحصیل را **در مراحل اولیه‌ی مسیر تحصیلی‌شان** شناسایی کنیم تا بتوان به‌موقع راهکارهای حمایتی برای آن‌ها اجرا کرد.

دیتاست شامل اطلاعاتی است که **در زمان ثبت‌نام دانشجو** (یا حداکثر تا پایان دو نیم‌سال اول) شناخته‌شده بوده‌اند، شامل مسیر تحصیلی، مشخصات جمعیت‌شناختی، و عوامل اجتماعی-اقتصادی. مسئله به‌صورت یک **طبقه‌بندی سه‌کلاسه** فرموله شده است: در پایان دوره‌ی عادی درسی، دانشجو در یکی از سه وضعیت زیر قرار می‌گیرد:

- **Dropout** (ترک تحصیل)
- **Enrolled** (هنوز مشغول تحصیل)
- **Graduate** (فارغ‌التحصیل شده)

### حمایت مالی
این دیتاست توسط برنامه‌ی **SATDAP - Capacitação da Administração Pública**، با گرنت شماره‌ی **POCI-05-5762-FSE-000191** در پرتغال، پشتیبانی مالی شده است.

### واحد مشاهده
هر ردیف از داده، نماینده‌ی یک **دانشجو** است.

### پیش‌پردازش داده
داده‌ها تحت یک فرایند پاکسازی دقیق قرار گرفته‌اند تا ناهنجاری‌ها، داده‌های پرت غیرقابل‌توضیح و مقادیر گمشده حذف شوند. در نتیجه، این دیتاست **هیچ مقدار گمشده‌ای ندارد**.

### تقسیم پیشنهادی داده
در پروژه‌ی اصلی، داده به نسبت **۸۰٪ آموزش و ۲۰٪ آزمون** تقسیم شده است.

### آمار کلی
- تعداد کل رکوردها (دانشجویان): **۴٬۴۲۴ نفر**
- تعداد ویژگی‌ها (Features): **۳۶ فیچر**
- تعداد متغیر هدف: **۱ متغیر (Target)** با سه کلاس
- توزیع کلاس‌های هدف: Graduate ۲٬۲۰۹ نفر (%۵۰) | Dropout ۱٬۴۲۱ نفر (%۳۲) | Enrolled ۷۹۴ نفر (%۱۸)

فیچرهای این دیتاست را می‌توان در هفت دسته‌ی اصلی طبقه‌بندی کرد: مشخصات فردی و جمعیت‌شناسی، وضعیت اقتصادی-اجتماعی، زمینه‌ی خانوادگی، مسیر ورود به دانشگاه، عملکرد تحصیلی نیم‌سال اول، عملکرد تحصیلی نیم‌سال دوم، و متغیرهای کلان‌اقتصادی زمان ثبت‌نام. جدول کامل این فیچرها به همراه توضیحات در ادامه آمده است.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.dpi"] = 120
```

```python
df = pd.read_excel("dataset.xlsx", sheet_name="data")

df.columns = [c.strip() for c in df.columns]

df.head()
```

## Q1. What is the distribution of the target classes?

```python


plt.figure(figsize=(6, 4))

order = df["Target"].value_counts().index

ax = sns.countplot(
    data=df,
    x="Target",
    order=order,
    palette="viridis"
)

for p in ax.patches:
    ax.annotate(
        f"{int(p.get_height())}",
        (p.get_x() + p.get_width()/2, p.get_height()),
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.title("Q1. Distribution of Target Classes")
plt.xlabel("Target")
plt.ylabel("Number of Students")
plt.tight_layout()

plt.show()
```

<div dir="rtl">
از مجموع ۴۴۲۴ دانشجو، ۲۲۰۹ نفر (حدود ۵۰٪) در گروه **Graduate**، تعداد ۱۴۲۱ نفر (حدود ۳۲٪) در گروه **Dropout** و ۷۹۴ نفر (حدود ۱۸٪) در گروه **Enrolled** قرار دارند. بنابراین، توزیع کلاس‌ها کاملاً متوازن نیست و کلاس **Enrolled** نسبت به دو کلاس دیگر نمونه‌های کمتری دارد. این عدم توازن باید در مرحله‌ی مدل‌سازی مورد توجه قرار گیرد؛ برای مثال می‌توان از روش‌هایی مانند **Class Weighting** یا در صورت مناسب بودن شرایط، **SMOTE** استفاده کرد.

</div>

## Q2. Does 2nd semester grade differ across target groups?


```python

plt.figure(figsize=(6, 4))

sns.boxplot(
    data=df,
    x="Target",
    y="Curricular units 2nd sem (grade)",
    order=order,
    palette="Set2"
)

plt.title("Q2. 2nd Semester Grade by Target Group")
plt.xlabel("Target")
plt.ylabel("2nd Sem Grade (0-20)")
plt.tight_layout()

plt.show()
```

<div dir="rtl">
تفاوت بین گروه‌ها در این نمودار بسیار چشمگیر است. میانه‌ی معدل نیم‌سال دوم در گروه‌های **Graduate** و **Enrolled** حدود ۱۲ تا ۱۳ است و پراکندگی نسبتاً محدودی دارند. در مقابل، گروه **Dropout** میانه‌ای نزدیک به صفر و پراکندگی بسیار بیشتری دارد؛ به‌طوری‌که مقادیر آن از صفر تا حدود ۱۸ مشاهده می‌شود.

این الگو نشان می‌دهد که بخش قابل‌توجهی از دانشجویان گروه Dropout در نیم‌سال دوم نمره‌ای ثبت نکرده‌اند که می‌تواند با ترک تحصیل زودهنگام آن‌ها مرتبط باشد. در میان نمودارهای این تحلیل، این متغیر یکی از واضح‌ترین تفاوت‌ها را بین گروه‌های Target نشان می‌دهد.

</div>

## Q5. Do debtor / tuition-fees-not-up-to-date students drop out more?


```python

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

for ax, col, label in zip(
    axes,
    ["Debtor", "Tuition fees up to date"],
    ["Debtor", "Tuition Fees Up to Date"]
):
    
    ct = pd.crosstab(
        df[col],
        df["Target"],
        normalize="index"
    ) * 100
    
    ct = ct[order]
    
    ct.plot(
        kind="bar",
        stacked=True,
        ax=ax,
        colormap="viridis"
    )
    
    ax.set_title(f"Target % by {label}")
    ax.set_xlabel(label + " (0=No, 1=Yes)")
    ax.set_ylabel("Percentage")
    ax.legend(
        title="Target",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

plt.suptitle("Q5. Financial Status vs Target (Stacked %)")
plt.tight_layout()

plt.show()
```

<div dir="rtl">
بین وضعیت مالی دانشجویان و Target تفاوت قابل‌توجهی مشاهده می‌شود. در میان دانشجویان **بدهکار**، حدود ۲۰٪ Graduate هستند، در حالی که این مقدار در میان دانشجویان غیر بدهکار حدود ۵۴٪ است. همچنین سهم Dropout در گروه بدهکار بیشتر است.

تفاوت در متغیر **Tuition fees up to date** حتی چشمگیرتر است. در میان دانشجویانی که وضعیت پرداخت شهریه‌ی آن‌ها به‌روز نیست، سهم Graduate بسیار پایین و سهم Dropout بیش از ۸۵٪ است.

بنابراین، در این داده‌ها وضعیت بدهی و پرداخت شهریه با Target ارتباط بسیار واضحی دارند. البته این نمودارها به‌تنهایی نشان‌دهنده‌ی رابطه‌ی علّی نیستند و نمی‌توان بر اساس آن‌ها نتیجه گرفت که وضعیت مالی مستقیماً باعث ترک تحصیل شده است.

</div>

## Q8. Is parents' education level related to student outcome?


```python

top_quals = df["Mother's qualification"].value_counts().nlargest(6).index

sub = df[
    df["Mother's qualification"].isin(top_quals)
].copy()

grad_rate = (
    sub.assign(
        is_grad=(sub["Target"] == "Graduate").astype(int)
    )
    .groupby("Mother's qualification")["is_grad"]
    .mean()
    .sort_values(ascending=False) * 100
)

plt.figure(figsize=(7, 4.5))

sns.barplot(
    x=grad_rate.index.astype(str),
    y=grad_rate.values,
    palette="flare"
)

plt.title(
    "Q8. Graduation Rate (%) by Mother's Qualification "
    "(Top 6 categories)"
)

plt.xlabel("Mother's Qualification (code)")
plt.ylabel("Graduation Rate (%)")
plt.tight_layout()

plt.show()
```

<div dir="rtl">
در میان شش دسته‌ی پرتکرار مربوط به تحصیلات مادر، نرخ فارغ‌التحصیلی از حدود ۲۴٪ تا ۵۹٪ متغیر است. بیشترین نرخ مربوط به **کد ۳۸** و کمترین نرخ مربوط به **کد ۳۴** است که در داده‌ها به وضعیت نامشخص یا Unknown اشاره دارد.

این تفاوت نشان می‌دهد که بین دسته‌های مختلف تحصیلات مادر، تفاوت‌هایی در نرخ فارغ‌التحصیلی مشاهده می‌شود. با این حال، برای تفسیر این رابطه باید توجه داشت که کدهای این متغیر صرفاً دسته‌های عددی هستند و نمی‌توان بدون بررسی دقیق معنای هر کد، فاصله‌ی بین آن‌ها را به‌عنوان یک مقیاس عددی در نظر گرفت.

## Q10. Which numeric features correlate most with each other / target?


```python

df_corr = df.copy()

df_corr["Target_num"] = df_corr["Target"].map({
    "Dropout": 0,
    "Enrolled": 1,
    "Graduate": 2
})

num_cols = [
    "Admission grade",
    "Age at enrollment",
    "Curricular units 1st sem (approved)",
    "Curricular units 1st sem (grade)",
    "Curricular units 2nd sem (approved)",
    "Curricular units 2nd sem (grade)",
    "Unemployment rate",
    "Inflation rate",
    "GDP",
    "Target_num"
]

corr = df_corr[num_cols].corr()

plt.figure(figsize=(8, 6.5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True
)

plt.title("Q10. Correlation Heatmap (Key Numeric Features & Target)")
plt.tight_layout()

plt.show()
```
